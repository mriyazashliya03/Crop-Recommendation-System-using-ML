IMPORTING AND FEATURE DISTRIBUTION 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Crop_recommendation.csv")

features = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']

print("Ready!")
```

```python
df[features].hist(bins=20, figsize=(14, 8), color='steelblue', edgecolor='black')
plt.suptitle('Feature Distributions', fontsize=16)
plt.tight_layout()
plt.savefig('feature_distributions.png')
plt.show()
```

CORRELATION HEATMAP AND CROP COUNT

```python
plt.figure(figsize=(9, 6))
sns.heatmap(df[features].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.savefig('correlation_heatmap.png')
plt.show()

plt.figure(figsize=(14, 5))
sns.countplot(data=df, x='label', order=df['label'].value_counts().index, palette='viridis')
plt.xticks(rotation=45, ha='right')
plt.title('Number of Samples per Crop')
plt.tight_layout()
plt.savefig('crop_distribution.png')
plt.show()
```
